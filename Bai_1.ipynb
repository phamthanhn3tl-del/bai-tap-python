```python

# Bài 1.1: Dự đoán kết quả các phép toán trong Python
x = 23
y = 6
print("x / y =", x / y)
print("x // y =", x // y)
print("x % y =", x % y)
print("x ** 2 =", x ** 2)
print("-x // y =", -x // y)
print("-x % y =", -x % y)
print("2 ** 3 ** 2 =", 2 ** 3 ** 2)
print("(2 ** 3) ** 2 =", (2 ** 3) ** 2)
```
