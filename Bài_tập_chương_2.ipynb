Bài 2.4.

(Dự đoán kết quả)  Hàm đệ quy dưới đây trả về cặp (giá trị, số lời gọi). Dự đoán kết quả của g(2468), g(0) và g(1999). Hàm g thực chất tính đại lượng gì của n?

```python
def g(n):
    if n == 0:
        return 0, 1
    v, c = g(n // 10)
    return n % 10 + v, c + 1

print(g(2468), g(0), g(1999))
```

Vẽ vết gọi 2468→246→24→2→0 và đếm lời gọi.

Bài 2.7.

(Đọc lỗi và sửa mã)  Chương trình tính 100/soˆˊ_bạn phải xử lý được cả đầu vào không phải số (abc) lẫn 0. Chạy mã với hai đầu vào đó, giải thích các lỗi còn sót (có hai lỗi khác nhau) và sửa.

```python
try:
    so_ban = int(input("Số bạn: "))
    tb = 100 / so_ban

except ValueError:
    print("Hãy nhập số nguyên!")

except ZeroDivisionError:
    print("Số bạn phải khác 0!")

else:
    print(f"Kết quả: {tb}")
```

(Bẫy logic)  Chương trình dưới đây muốn in “x là 1 hoặc 2” khi x∈{1,2}, nhưng với x=5 vẫn in như vậy. Giải thích bằng quy tắc ưu tiên toán tử và giá trị “đúng/sai” của số, rồi viết hai cách sửa đúng.

```python
x = 5

if x == 1 or x == 2:
    print("x là 1 hoặc 2")
else:
    print("x khác 1 và 2")
```

```python
x = 5

if x in (1, 2):
    print("x là 1 hoặc 2")
else:
    print("x khác 1 và 2")
```

Bài 2.11.

(Bẫy logic)  Hàm tao_nhom cho mỗi lần gọi một nhóm mới chứa tên truyền vào. Dự đoán rồi chạy; vì sao hai nhóm lại giống nhau? Sửa hàm để a và b là hai danh sách độc lập.

```python
def tao_nhom(ten, thanh_vien=None):
    if thanh_vien is None:
        thanh_vien = []

    thanh_vien.append(ten)
    return thanh_vien

a = tao_nhom("An")
b = tao_nhom("Bình")

print(a, b, a is b)
```
