TH 4.1.
Viết lớp HinhCau có bán kính ban_kinh, các phương thức dien_tich() (4πr2), the_tich() (43πr3) và __str__. Hàm tạo phải báo ValueError nếu bán kính không dương. Kiểm tra với r=2 (diện tích 16π≈50,265, thể tích 32π3≈33,510) và r=−1.

```python

import math

class HinhCau:
    def __init__(self, ban_kinh):
        if ban_kinh <= 0:
            raise ValueError("Bán kính phải là số dương!")

        self.ban_kinh = ban_kinh

    def dien_tich(self):
        return 4 * math.pi * self.ban_kinh**2

    def the_tich(self):
        return (4/3) * math.pi * self.ban_kinh**3

    def __str__(self):
        return f"Hình cầu có bán kính: {self.ban_kinh}"

# Kiểm tra với r = 2
hinh_cau = HinhCau(2)

print(hinh_cau)
print("Diện tích:", hinh_cau.dien_tich())
print("Thể tích:", hinh_cau.the_tich())

# Kiểm tra với r = -1
try:
    hinh_loi = HinhCau(-1)
except ValueError as e:
    print("Lỗi:", e)
```

TH 4.3.
Viết lớp TaiKhoan (chủ tài khoản, số dư ban đầu mặc định 0 và không âm) với nap(tien), rut(tien), danh sách lich_su riêng từng tài khoản. Định nghĩa ngoại lệ SoDuKhongDu (kế thừa Exception) và dùng nó khi rút quá số dư; số tiền không dương thì báo ValueError. Kiểm tra bất biến “số dư ≥0” (Mệnh đề 4.1) với dãy: số dư đầu 1000000, nạp 500000, rút 200000, rút 2000000.

```python

# Định nghĩa ngoại lệ số dư không đủ
class SoDuKhongDu(Exception):
    pass


class TaiKhoan:
    def __init__(self, chu_tai_khoan, so_du=0):
        if so_du < 0:
            raise ValueError("Số dư ban đầu không được âm!")

        self.chu_tai_khoan = chu_tai_khoan
        self.so_du = so_du
        self.lich_su = []

    # Phương thức nạp tiền
    def nap(self, tien):
        if tien <= 0:
            raise ValueError("Số tiền nạp phải lớn hơn 0!")

        self.so_du += tien
        self.lich_su.append(f"Nạp: +{tien}")

    # Phương thức rút tiền
    def rut(self, tien):
        if tien <= 0:
            raise ValueError("Số tiền rút phải lớn hơn 0!")

        if tien > self.so_du:
            raise SoDuKhongDu("Số dư không đủ để thực hiện giao dịch!")

        self.so_du -= tien
        self.lich_su.append(f"Rút: -{tien}")

    def __str__(self):
        return f"Chủ tài khoản: {self.chu_tai_khoan}, Số dư: {self.so_du:,} VNĐ"


# Kiểm tra chương trình
tk = TaiKhoan("Nguyen Van A", 1000000)

print("Ban đầu:", tk)

# Nạp 500.000
tk.nap(500000)
print("Sau khi nạp:", tk)

# Rút 200.000
tk.rut(200000)
print("Sau khi rút:", tk)

# Rút 2.000.000
try:
    tk.rut(2000000)
except SoDuKhongDu as e:
    print("Lỗi:", e)

# In lịch sử giao dịch
print("\nLịch sử giao dịch:")
for giao_dich in tk.lich_su:
    print(giao_dich)

# Kiểm tra bất biến số dư >= 0
assert tk.so_du >= 0
print("\nKiểm tra bất biến số dư >= 0:", tk.so_du >= 0)
```

TH 4.5.
Thiết kế lớp cha HinhHoc có phương thức dien_tich() chỉ báo NotImplementedError (buộc lớp con phải cài đặt) và __str__; ba lớp con HinhTron(r), HinhChuNhat(a, b), HinhTamGiac(a, b, c) (công thức Heron). Với danh sách gồm hình tròn r=1, chữ nhật 3×4 và tam giác (3,4,5): in các hình theo diện tích tăng dần và tổng diện tích (phải bằng π+18), nhờ đa hình.

```python

import math

# Lớp cha
class HinhHoc:
    def dien_tich(self):
        raise NotImplementedError("Lớp con phải cài đặt phương thức dien_tich()")

    def __str__(self):
        return "Hình học"


# Lớp hình tròn
class HinhTron(HinhHoc):
    def __init__(self, r):
        self.r = r

    def dien_tich(self):
        return math.pi * self.r**2

    def __str__(self):
        return f"Hình tròn (r={self.r})"


# Lớp hình chữ nhật
class HinhChuNhat(HinhHoc):
    def __init__(self, a, b):
        self.a = a
        self.b = b

    def dien_tich(self):
        return self.a * self.b

    def __str__(self):
        return f"Hình chữ nhật ({self.a} x {self.b})"


# Lớp hình tam giác
class HinhTamGiac(HinhHoc):
    def __init__(self, a, b, c):
        if (a + b <= c or a + c <= b or b + c <= a):
            raise ValueError("Ba cạnh không tạo thành tam giác!")

        self.a = a
        self.b = b
        self.c = c

    def dien_tich(self):
        p = (self.a + self.b + self.c) / 2
        return math.sqrt(
            p * (p - self.a) * (p - self.b) * (p - self.c)
        )

    def __str__(self):
        return f"Hình tam giác ({self.a}, {self.b}, {self.c})"


# Tạo danh sách các hình
ds_hinh = [
    HinhTron(1),
    HinhChuNhat(3, 4),
    HinhTamGiac(3, 4, 5)
]

# Sắp xếp theo diện tích tăng dần
ds_hinh.sort(key=lambda h: h.dien_tich())

# In danh sách hình
print("Danh sách hình theo diện tích tăng dần:")

for h in ds_hinh:
    print(f"{h}: Diện tích = {h.dien_tich():.3f}")

# Tính tổng diện tích
tong = sum(h.dien_tich() for h in ds_hinh)

print(f"\nTổng diện tích = {tong:.3f}")
print(f"Giá trị chính xác = π + 18 = {math.pi + 18:.3f}")

# Kiểm tra kết quả
assert math.isclose(tong, math.pi + 18)

print("Kiểm tra tổng diện tích: ĐÚNG!")
```

TH 4.7.
Với tệp data/goldprice.csv (giá vàng theo tháng, cột Date dạng yyyy-mm, Price), tính giá trung bình của từng năm đủ 12 tháng, rồi mức tăng (%) so với năm liền trước. In ba năm tăng mạnh nhất và năm giảm mạnh nhất. (Đáp số biết trước: năm 1980 tăng mạnh nhất, năm 1981 giảm mạnh nhất.)

```python

import pandas as pd

# 1. Đọc dữ liệu CSV
df = pd.read_csv("goldprice.csv")

# 2. Chuyển cột Date sang kiểu datetime
df["Date"] = pd.to_datetime(df["Date"], format="%Y-%m")

# 3. Lấy năm từ cột Date
df["Year"] = df["Date"].dt.year

# 4. Tính giá trung bình và số tháng của từng năm
thong_ke = df.groupby("Year").agg(
    gia_tb=("Price", "mean"),
    so_thang=("Price", "count")
)

# 5. Chỉ giữ những năm có đủ 12 tháng
thong_ke = thong_ke[thong_ke["so_thang"] == 12]

# 6. Tính phần trăm tăng trưởng so với năm liền trước
thong_ke["tang_truong"] = (
    thong_ke["gia_tb"].pct_change() * 100
)

# 7. Chỉ giữ các năm có năm liền trước
# cũng nằm trong danh sách năm đủ 12 tháng
thong_ke = thong_ke.dropna(subset=["tang_truong"])

# 8. In kết quả thống kê
print("THỐNG KÊ GIÁ VÀNG TRUNG BÌNH THEO NĂM:")
print(thong_ke)

# 9. Ba năm tăng mạnh nhất
top_3 = thong_ke.nlargest(3, "tang_truong")

print("\n3 NĂM TĂNG GIÁ MẠNH NHẤT:")
print(top_3[["gia_tb", "tang_truong"]])

# 10. Năm giảm mạnh nhất
nam_giam = thong_ke.nsmallest(1, "tang_truong")

print("\nNĂM GIẢM GIÁ MẠNH NHẤT:")
print(nam_giam[["gia_tb", "tang_truong"]])
```

Viết chương trình đọc tệp bằng csv.DictReader: dòng nào không đổi được sang số thì bị bỏ qua và ghi lại số thứ tự dòng trong tệp (dòng tiêu đề là dòng 1). Dùng ngoại lệ tự định nghĩa DongKhongHopLe. In các dòng hợp lệ, danh sách dòng lỗi và lương trung bình của các dòng hợp lệ.

```python
import pandas as pd

data = {
    "nam_kn": [1, 2, "abc", 3, 4, 5],
    "luong": [30, 35, 40, None, 50, 55]
}

df = pd.DataFrame(data)

df.to_csv("luong.csv", index=False)

print("Đã tạo file luong.csv")
```

```python

import csv

# Định nghĩa ngoại lệ riêng
class DongKhongHopLe(Exception):
    pass


dong_hop_le = []
dong_loi = []
tong_luong = 0

# Đọc dữ liệu CSV
with open("luong.csv", "r", encoding="utf-8") as file:
    reader = csv.DictReader(file)

    for so_dong, row in enumerate(reader, start=2):
        try:
            nam_kn = int(row["nam_kn"])
            luong = float(row["luong"])

            # Kiểm tra dữ liệu hợp lệ
            if nam_kn < 0 or luong < 0:
                raise DongKhongHopLe("Dữ liệu không được âm!")

            dong_hop_le.append({
                "nam_kn": nam_kn,
                "luong": luong
            })

            tong_luong += luong

        except (ValueError, TypeError, DongKhongHopLe, KeyError):
            dong_loi.append(so_dong)

# In các dòng hợp lệ
print("CÁC DÒNG HỢP LỆ:")
for row in dong_hop_le:
    print(row)

# In danh sách dòng lỗi
print("\nDANH SÁCH DÒNG LỖI:")
print(dong_loi)

# Tính lương trung bình
if dong_hop_le:
    luong_tb = tong_luong / len(dong_hop_le)
    print(f"\nLương trung bình: {luong_tb:.2f}")
else:
    print("Không có dữ liệu hợp lệ!")
```
