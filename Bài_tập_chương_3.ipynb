Bài 3.4.

(Dự đoán kết quả)  Dự đoán đầu ra (chú ý get so với [ ], và các phép toán tập hợp). Với dòng cuối, hãy nhớ tập hợp không có thứ tự nên chương trình dùng sorted.

```python
A, B = {1, 2, 3, 4}, {3, 4, 5}
print(A & B, A | B, A - B, B - A, A ^ B)
print(len(A | B) == len(A) + len(B) - len(A & B))
diem = {"An": 8, "Binh": 7}
print(diem.get("An"), diem.get("Chi"), diem.get("Chi", 0))
diem["Chi"] = diem.get("Chi", 0) + 5
print(sorted(diem), len(diem))
print(sorted({x % 3 for x in range(10)}), {c: len(c) for c in ["ab", "cde"]})
```

Bài 3.6.

(Đọc lỗi và sửa mã) [AI] Công cụ AI sinh hàm thêm điểm vào danh sách như dưới đây, nhưng sau khi gọi thì danh sách ban đầu không đổi (kết quả mong đợi [5, 6, 9]). Giải thích vì sao, rồi sửa theo hai cách: (i) hàm sửa trực tiếp danh sách được truyền vào; (ii) hàm trả về danh sách mới và nơi gọi gán lại kết quả.

```python
def them_diem(ds, diem):
    ds = ds + [diem]
    return ds

danh_sach = [5, 6]
danh_sach = them_diem(danh_sach, 9)
print(danh_sach)
```

Bài 3.8.

(Bẫy logic)  Mã dưới đây muốn xoá mọi số chẵn khỏi danh sách (kết quả mong đợi [1, 7]). Chạy xem kết quả, giải thích vì sao một số chẵn vẫn còn, rồi sửa bằng hai cách: list comprehension, và duyệt trên một bản sao.

```python
ds = [1, 2, 4, 6, 7, 8]

for x in ds[:]:
    if x % 2 == 0:
        ds.remove(x)

print(ds)
```
