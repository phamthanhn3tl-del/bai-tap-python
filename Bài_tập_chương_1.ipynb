TH 1.1.
Tính chu vi và diện tích tam giác biết ba cạnh a,b,c bằng công thức Heron S=p(p−a)(p−b)(p−c), p=(a+b+c)/2. Kiểm tra với (13,14,15): diện tích bằng 84; và với (3,4,5): diện tích bằng 6.

```python

import math

def tinh_tam_giac(a, b, c):
    if a <= 0 or b <= 0 or c <= 0:
        print("Ba cạnh phải là số dương.")
        return

    if a + b <= c or a + c <= b or b + c <= a:
        print("Ba cạnh không tạo thành tam giác.")
        return
    P = a + b + c
    p = P / 2
    S = math.sqrt(p * (p - a) * (p - b) * (p - c))

    print(f"Ba cạnh: a = {a}, b = {b}, c = {c}")
    print(f"Chu vi tam giác: {P}")
    print(f"Nửa chu vi: {p}")
    print(f"Diện tích tam giác: {S:g}")
    print("-" * 40)
tinh_tam_giac(13, 14, 15)
tinh_tam_giac(3, 4, 5)
```

Gửi tiết kiệm kỳ hạn 1 tháng số tiền 50000000 đồng với lãi suất 6%/năm; mỗi tháng lãi (bằng 6%/12 của số dư) được cộng vào gốc. In số dư sau 1, 2 và 3 tháng, làm tròn đến đồng, có dấu ngăn cách hàng nghìn.

```python
tien_gui = 50000000
lai_suat = 0.06 / 12
for thang in range(1, 4):
    tien_gui = tien_gui * (1 + lai_suat)
    print(f"Số dư sau {thang} tháng: {tien_gui:,.0f} đồng")
```

TH 1.3.
Đổi số phút tong_phut thành dạng “x ngày y giờ z phút”. Kiểm tra với 5000 phút (phải được 3 ngày 11 giờ 20 phút), 1440 phút và 59 phút.

```python
def doi_phut(tong_phut):
    ngay = tong_phut // 1440
    gio = (tong_phut % 1440) // 60
    phut = tong_phut % 60

    print(f"{tong_phut} phút = {ngay} ngày {gio} giờ {phut} phút")
doi_phut(5000)
doi_phut(1440)
doi_phut(59)
```

TH 1.4.
Một máy rút tiền cần trả so_tien = 1_880_000 đồng bằng các tờ mệnh giá 500000, 200000, 100000, 50000, 20000, 10000 đồng, ưu tiên tờ mệnh giá lớn trước. In số tờ của từng mệnh giá và số tiền còn lại (nếu có). Kiểm tra: tổng giá trị các tờ đã trả phải bằng số tiền ban đầu.

```python
so_tien = 1_880_000
tien_ban_dau = so_tien
menh_gia = [500000, 200000, 100000, 50000, 20000, 10000]
tong_da_tra = 0
for menh in menh_gia:
    so_to = so_tien // menh
    so_tien = so_tien % menh

    gia_tri = so_to * menh
    tong_da_tra += gia_tri

    print(f"Mệnh giá {menh:,} đồng: {so_to} tờ")
print(f"Số tiền còn lại: {so_tien:,} đồng")
print(f"Tổng tiền đã trả: {tong_da_tra:,} đồng")
print("Kiểm tra:", tong_da_tra == tien_ban_dau)
```

TH 1.5.
Cho số nguyên bốn chữ số n = 7539. Chỉ dùng // và % (không dùng chuỗi) để tách các chữ số hàng nghìn, trăm, chục, đơn vị; in từng chữ số, tổng các chữ số và số đảo ngược (9357). Kiểm tra lại số đảo ngược bằng cách dùng chuỗi: int(str(n)[::-1]).

```python
n = 7539
hang_nghin = n // 1000
hang_tram = (n // 100) % 10
hang_chuc = (n // 10) % 10
hang_don_vi = n % 10
tong = hang_nghin + hang_tram + hang_chuc + hang_don_vi
so_dao_nguoc = (hang_don_vi * 1000 +
                hang_chuc * 100 +
                hang_tram * 10 +
                hang_nghin)
kiem_tra = int(str(n)[::-1])
print("Hàng nghìn:", hang_nghin)
print("Hàng trăm:", hang_tram)
print("Hàng chục:", hang_chuc)
print("Hàng đơn vị:", hang_don_vi)
print("Tổng các chữ số:", tong)
print("Số đảo ngược:", so_dao_nguoc)
print("Kiểm tra bằng chuỗi:", kiem_tra)
print("Kết quả kiểm tra:", so_dao_nguoc == kiem_tra)
```
